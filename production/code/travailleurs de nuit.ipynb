```python
import pandas as pd
```

```python
# Import des données de population par CSP (source INSEE)
path_data_pop =  r"C:\Users\9920510N\SNCF\Datalab' Mass Transit GrpO365 - Documents\Rayonnement\Datathon\2026\Hackaviz IdFM\Données\Données_POP_CSP.xlsx"
data_pop = pd.read_excel(path_data_pop,sheet_name="COM_2022") #On prend le dernier rescencement communal
```

```python
# Nettoyage des données

#On supprime les premières lignes inutiles
data_pop = data_pop.iloc[13:,:].reset_index(drop=True)

# La première ligne devient les noms de colonnes
data_pop.columns = data_pop.iloc[0]

# On supprime cette ligne des données et on réindexe
data_pop = data_pop[2:].reset_index(drop=True)
```

```python
display(data_pop.head(10))
```

```python
# Restriction aux communes d'Ile-de-France
data_idf = data_pop[data_pop["Département\nen géographie courante"].isin(["75","77","78","91","92","93","94","95"])]

#Création de la colonne Code_commune_INSEE
data_idf["code_commune_INSEE"] = data_idf["Département\nen géographie courante"]+data_idf["Commune\nen géographie courante"]
```

```python
liste_colonnes_utiles = ["Ouvriers\nActifs ayant un emploi\nRP2022",
                         "Employés\nActifs ayant un emploi\nRP2022",
                         "Professions intermédiaires\nActifs ayant un emploi\nRP2022",
                         "Cadres et professions intellectuelles supérieures\nActifs ayant un emploi\nRP2022",
                         "Libellé de commune",
                         "code_commune_INSEE"]

#Création du dataframe d'étude
data = data_idf[liste_colonnes_utiles]
data = data.rename(columns = {"Ouvriers\nActifs ayant un emploi\nRP2022" : "Ouvriers",
                              "Employés\nActifs ayant un emploi\nRP2022" : "Employes",
                              "Professions intermédiaires\nActifs ayant un emploi\nRP2022" : "Inters",
                              "Cadres et professions intellectuelles supérieures\nActifs ayant un emploi\nRP2022" : "Cadres"})
display(data.head(10))
```

```python
#Methode

#1 - Calculer des % par CSP en fonction de la population sondée lors de l'enquête
#2 - Utiliser le % à la hausse issu de l'enquête INSEE sur les horaires atypiques de travail pour déterminer le nombre de travailleurs de nuit

data["Ouvriers_nuit"] = data["Ouvriers"]*0.15
data["Employes_nuit"] = data["Employes"]*0.1
data["Inters_nuit"] = data["Inters"]*0.1
data["Cadres_nuit"] = data["Cadres"]*0.08

data["Reference_pop"] = data["Ouvriers"].astype(float)+data["Employes"].astype(float)+data["Inters"].astype(float)+data["Cadres"].astype(float)
data["Travailleurs_nuit"] = data["Ouvriers_nuit"].astype(float)+data["Employes_nuit"].astype(float)+data["Inters_nuit"].astype(float)+data["Cadres_nuit"].astype(float)
data["Travailleurs_nuit_%"] = data["Travailleurs_nuit"]/data["Reference_pop"]*100
```

```python
# Création du dataset final
data_final = data[["code_commune_INSEE","Libellé de commune","Travailleurs_nuit_%"]]
#data_final.to_excel(r"C:\Users\9920510N\SNCF\Datalab' Mass Transit GrpO365 - Documents\Rayonnement\Datathon\2026\Hackaviz IdFM\Données\Résultats_travail_nuit.xlsx")
```

```python
#Ajout de l'importance des TC dans les flux-domicile-travail
path_TC = r"C:\Users\9920510N\SNCF\Datalab' Mass Transit GrpO365 - Documents\Rayonnement\Datathon\2026\Hackaviz IdFM\Données\Resultats_DT_transport.csv"   # fichier de résultat
resultat = pd.read_csv(path_TC, sep=";", decimal=",", encoding="utf-8-sig")

#On vient fusionne rles deux bases sur le code_commune_INSEE
data_merged = data_final.merge(resultat[["geocode_commune","%_Transports en commun"]],right_on="geocode_commune",left_on="code_commune_INSEE",how="left")
data_merged = data_merged.drop("geocode_commune",axis=1)

# Ajout de la colonne "Travailleurs_nuit" pondéré par le % de prise des TC dans les flux domicile-travail
data_merged["Travailleurs_nuit_%_pond"] = data_merged["Travailleurs_nuit_%"]*data_merged["%_Transports en commun"]/100
data_merged.to_excel(r"C:\Users\9920510N\SNCF\Datalab' Mass Transit GrpO365 - Documents\Rayonnement\Datathon\2026\Hackaviz IdFM\Données\Résultats_travail_nuit.xlsx")
display(data_merged)
```

```python
#Analyse rapide des résultats
import pandas as pd

data_final = data[["code_commune_INSEE", "Libellé de commune", "Travailleurs_nuit_%"]].copy()

# 1. Typage (important si les en-têtes viennent de la 1re ligne : les colonnes sont souvent en object)
data_final["code_commune_INSEE"] = data_final["code_commune_INSEE"].astype(str).str.zfill(5)  # garde les 0 initiaux et la Corse (2A/2B)
data_final["Travailleurs_nuit_%"] = pd.to_numeric(
    data_final["Travailleurs_nuit_%"].astype(str).str.replace(",", ".").str.replace("%", ""),
    errors="coerce"
)

col = "Travailleurs_nuit_%"

# 2. Qualité des données
print("Nombre de lignes :", len(data_final))
print("Communes uniques (code INSEE) :", data_final["code_commune_INSEE"].nunique())
print("Codes INSEE en doublon :", data_final["code_commune_INSEE"].duplicated().sum())
print("Valeurs manquantes :\n", data_final.isna().sum())

# 3. Statistiques descriptives principales
stats = data_final[col].describe(percentiles=[.1, .25, .5, .75, .9]).to_frame().T
stats["variance"] = data_final[col].var()
stats["asymetrie"] = data_final[col].skew()
stats["aplatissement"] = data_final[col].kurt()
stats["coef_variation_%"] = data_final[col].std() / data_final[col].mean() * 100
print(stats.round(2).T)

# 4. Top / Flop 10 des communes
print("\nTop 10 :\n", data_final.nlargest(10, col)[["Libellé de commune", col]])
print("\nFlop 10 :\n", data_final.nsmallest(10, col)[["Libellé de commune", col]])

# 5. Répartition par tranches
data_final["tranche"] = pd.cut(data_final[col], bins=[0, 5, 10, 15, 20, 100], include_lowest=True)
print("\nRépartition par tranche :\n", data_final["tranche"].value_counts().sort_index())

# 6. Valeurs atypiques (méthode IQR)
q1, q3 = data_final[col].quantile([.25, .75])
iqr = q3 - q1
outliers = data_final[(data_final[col] < q1 - 1.5 * iqr) | (data_final[col] > q3 + 1.5 * iqr)]
print(f"\nValeurs atypiques : {len(outliers)} communes")

# 7. Bonus : agrégation par département (2 premiers caractères du code INSEE ; 3 pour les DOM, à affiner)
data_final["departement"] = data_final["code_commune_INSEE"].str[:2]
print(data_final.groupby("departement")[col].agg(["count", "mean", "median", "min", "max"]).round(2))
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
data_final[col].plot.hist(bins=40, ax=ax[0], title="Distribution du % de travailleurs de nuit")
data_final[col].plot.box(ax=ax[1], title="Boîte à moustaches")
plt.tight_layout()
plt.show()
```
